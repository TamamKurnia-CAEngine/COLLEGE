# Analisis Perbandingan Fungsi Aktivasi pada MLP dengan Dataset Iris
Notebook ini membandingkan performa arsitektur Multi-Layer Perceptron (MLP) menggunakan tiga fungsi aktivasi yang berbeda: **Sigmoid**, **Tanh**, dan **ReLU**.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# 1. Memuat Dataset
df = pd.read_csv('Iris.csv')
if 'Id' in df.columns:
    df = df.drop(columns=['Id'])

X = df.drop(columns=['Species'])
y = df['Species']

# 2. Preprocessing
le = LabelEncoder()
y_encoded = le.fit_transform(y)

X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.3, random_state=42, stratify=y_encoded)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 3. Training dan Evaluasi MLP
activations = ['logistic', 'tanh', 'relu']
results = []

for act in activations:
    mlp = MLPClassifier(hidden_layer_sizes=(10, 10), activation=act, random_state=42, max_iter=2000)
    mlp.fit(X_train_scaled, y_train)
    y_pred = mlp.predict(X_test_scaled)
    
    acc = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred, average='macro')
    rec = recall_score(y_test, y_pred, average='macro')
    f1 = f1_score(y_test, y_pred, average='macro')
    
    display_name = 'Sigmoid' if act == 'logistic' else ('Tanh' if act == 'tanh' else 'ReLU')
    results.append({
        'Fungsi Aktivasi': display_name,
        'Accuracy': acc,
        'Precision': prec,
        'Recall': rec,
        'F1-Score': f1
    })

df_results = pd.DataFrame(results)
print(df_results.to_string(index=False))
```

## Kesimpulan

Berdasarkan hasil pengujian Multi-Layer Perceptron (MLP) pada dataset Iris menggunakan tiga fungsi aktivasi berbeda, dapat disimpulkan bahwa:

1. **Fungsi Aktivasi Tanh** memberikan performa terbaik dan paling optimal dengan nilai seluruh metrik (*Accuracy, Precision, Recall, F1-Score*) mencapai **95,56%**[cite: 1].
2. **Fungsi Aktivasi Sigmoid** berada di posisi kedua dengan performa yang cukup solid pada tingkat akurasi **93,33%**[cite: 1].
3. **Fungsi Aktivasi ReLU** menghasilkan performa terendah pada arsitektur ini dengan akurasi **91,11%**[cite: 1].

**Rekomendasi:** Untuk klasifikasi dataset Iris dengan arsitektur MLP ini, fungsi aktivasi **Tanh** adalah pilihan terbaik karena menghasilkan tingkat pengenalan model yang paling tinggi dan seimbang[cite: 1].
