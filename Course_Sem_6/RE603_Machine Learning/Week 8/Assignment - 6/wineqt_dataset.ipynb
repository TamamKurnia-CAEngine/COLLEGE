## Prepare

```python
import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

sns.set_style('whitegrid')
```

## Load Our Dataset

```python
df = pd.read_csv('WineQT.csv')

df.head()
```

### Melihat Informasi Dataset

```python
print(df.shape)

df.info()
```

## Exploratory Data Analysis (EDA)

### Statistik Deskriptif

```python
df.describe()
```

### Cek Missing Value

```python
df.isnull().sum()
```

### Distribusi Label Quality

```python
plt.figure(figsize=(8,5))

sns.countplot(data=df, x='quality')

plt.title('Distribusi Quality Wine')
plt.show()
```

### Correlation Heatmap

```python
plt.figure(figsize=(12,8))

sns.heatmap(df.corr(), cmap='coolwarm')

plt.title('Correlation Heatmap')
plt.show()
```

## Feature Engineering

### Menghapus Duplicate

```python
df = df.drop_duplicates().reset_index(drop=True)
```

### Menghapus Kolom ID

```python
df = df.drop(columns=['Id'])
```

### Menentukan Feature dan Label

```python
X = df.drop(columns=['quality'])

y = df['quality']
```

### Feature Scaling

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

## K-means Clustering

### Membuat Model K-Means

```python
kmeans = KMeans(n_clusters=3, random_state=42)

cluster = kmeans.fit_predict(X_scaled)
```

### Menambahkan Hasil Cluster

```python
df['cluster'] = cluster
```

## Metode Elbow

```python
inertia = []

for i in range(1,11):

    kmeans = KMeans(n_clusters=i, random_state=42)
    kmeans.fit(X_scaled)

    inertia.append(kmeans.inertia_)
```

### Visualisasi Elbow Method

```python
plt.figure(figsize=(8,5))

plt.plot(range(1,11), inertia, marker='o')

plt.xlabel('Jumlah Cluster')
plt.ylabel('Inertia')
plt.title('Elbow Method')

plt.show()
```

## Bandingkan Dengan Label Anotator

### Visualisasi Cluster

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    x=df['alcohol'],
    y=df['volatile acidity'],
    hue=df['cluster']
)

plt.title('Cluster K-Means')
plt.show()
```

### Visualisasi Quality Asli

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    x=df['alcohol'],
    y=df['volatile acidity'],
    hue=df['quality']
)

plt.title('Quality Asli')
plt.show()
```

## Via Score Plot

### Menghitung Silhouette Score

```python
sil_score = silhouette_score(X_scaled, cluster)

print(sil_score)
```

## Evaluasi Pada K-Means Menggunakan Via Score Secara Visualisasi

### Visualisasi Silhouette Evaluation

```python
from sklearn.metrics import silhouette_samples

silhouette_vals = silhouette_samples(X_scaled, cluster)

y_lower = 10

plt.figure(figsize=(10,6))

for i in range(3):

    cluster_silhouette_vals = silhouette_vals[cluster == i]
    cluster_silhouette_vals.sort()

    y_upper = y_lower + len(cluster_silhouette_vals)

    plt.fill_betweenx(
        np.arange(y_lower, y_upper),
        0,
        cluster_silhouette_vals
    )

    y_lower = y_upper + 10

plt.axvline(sil_score, color='red', linestyle='--')

plt.title('Silhouette Plot')
plt.xlabel('Silhouette Coefficient')
plt.ylabel('Cluster')

plt.show()
```
